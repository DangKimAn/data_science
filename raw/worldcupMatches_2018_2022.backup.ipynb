```python
def transform_times(time):
    time = time.split(' ')
    time = ' '.join(time[1:])
    time = time.split(' - ')
    new_time =  ' '.join(x.strip() for x in time)
    new_time = new_time.replace('UTC' , '')

    if '+' in new_time:
        st, nd = new_time.split('+')
        if int(nd) < 10:
            new_time = st.strip()+':00+0'+nd+':00'
        else:
            new_time = st.strip()+':00+'+nd+':00'
    else:
        st, nd = new_time.split('-')

        if int(nd) < 10:
            new_time = st.strip()+':00-0'+nd+':00'
        else:
            new_time = st.strip()+':00-'+nd+':00'
    times = new_time.split(' ')
    new_time = f"{times[2]}-{times[0]}-{times[1]} {times[3]}"
    return new_time
print(transform_times('Thu 06 14 2018 - 18:00 UTC+3'))
print(transform_times('Sat 07 18 2026 - 17:00 UTC-4'))
```

```python
def handle_wincondition(ip):
    out =[]

    pattern = r"(\d+)-(\d+),\s(\d+)-(\d+)"
    match = re.search(pattern, ip)

    if match:
        for x in list(match.groups()):
            out.append(int(x))         
    else:
        out.append(None)
        out.append(None)
        out.append(None)
        out.append(None)
        
    pattern = r"(\d+)-(\d+) pen."
    match = re.search(pattern, ip)

    if match:
        for x in list(match.groups()):
            out.append(int(x))        
    else:
        out.append(None)
        out.append(None)
        
    
    return out


print(handle_wincondition( 'a.e.t., 3-0 pen.'))
```

```python
columns = ['Home First Extra Time Goals' ,'Away First Extra Time Goals' , 'Home Second Extra Time Goals',
         'Away Second Extra Time Goals' , 'Home Penalties Goals' , 'Away Penalties Goals']
# columns = ['Home Penalties Goals' , 'Away Penalties Goals']
extra_time_goals_df =  pd.DataFrame(columns= columns, data= df['Win conditions'].apply(handle_wincondition).to_list(),
                               index= df.index )
extra_time_goals_df
```

```python
# Change roundID of match "Match for thirds place" in 2026 from 202660 to 202650

# fix for more, dont use stage because stage will be changed 
# df.loc[(df['RoundID'] == 202660) & (df['Stage'] == 'Match for third place'), 'RoundID'] = 202650
# df.groupby('RoundID')['Year'].count()

def fix_third_place_roundid(df, target_id, new_id, date_col='Datetime'):
    """
    Fix the round ID for the 3rd place match using the match date.
    
    Args:
    - df: The data table.
    - target_id: The ID that has both matches (example: 202660).
    - new_id: The new ID for the 3rd place match (example: 202650).
    - date_col: The name of the date column.
    """
    # 1. Find all rows with the target_id
    idx_target = df[df['RoundID'] == target_id].index

    # 2. Check if we have more than 1 match
    if len(idx_target) > 1:
        # Find the first match (the earliest date)
        idx_third_place = df.loc[idx_target, date_col].idxmin()
        
        # Change the ID to the new_id
        df.loc[idx_third_place, 'RoundID'] = new_id
        print(f"Success! Changed 3rd place ID: {target_id} -> {new_id}.")
        
    elif len(idx_target) == 1:
        # Only 1 match found. We missed the 3rd place match.
        print(f"Only 1 match found for {target_id}. No changes made.")
        
    else:
        # No matches found
        print(f"No data found for {target_id}.")
        
    return df

# How to use this function:
df = fix_third_place_roundid(df, target_id=202660, new_id=202650)
df.groupby('RoundID')['Year'].count()
```
