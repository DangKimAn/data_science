```python
# check name team 

def team_name_inside(arr ):
    teams = arr
    related_teams = {}
    for team1 in teams:
        for team2 in teams:
            # Nếu tên team1 nằm trong tên team2 (và hai tên không giống hệt nhau)
            if team1 != team2 and team1 in team2:
                related_teams[team2] = team1
    return related_teams
```

```python
# Handle team name duplicates 
df['Home Team Name'] = df['Home Team Name'].replace(
    team_name_inside(df['Home Team Name'].unique())
)
df['Home Team Name'].unique()
```
