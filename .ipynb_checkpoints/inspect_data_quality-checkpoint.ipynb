# FIFA World Cup 2026 - Data Quality & Profiling Notebook
Notebook này giúp bạn trực tiếp kiểm tra, đánh giá và khám phá chất lượng của bộ dữ liệu  trước khi bàn giao cho nhóm vẽ biểu đồ.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Cấu hình giao diện biểu đồ
sns.set_theme(style='whitegrid')
plt.rcParams['font.sans-serif'] = 'DejaVu Sans'
plt.rcParams['figure.dpi'] = 120

print('Libraries loaded successfully!')
```

## 1. Tải và Khám Phá Tổng Quan Bộ Dữ Liệu

```python
df = pd.read_csv('worldcup_master_cleaned.csv')
dim_tourn = pd.read_csv('dim_tournaments.csv')
dim_stad = pd.read_csv('dim_stadiums.csv')
dim_conf = pd.read_csv('dim_confederations.csv')

print(f'Số dòng: {len(df):,} trận đấu')
print(f'Số cột : {len(df.columns)} cột tính năng')
display(df.head(3))
```

## 2. Kiểm Tra Giá Trị Thiếu (Missing Values / Completeness)
Kiểm tra xem 18 cột phân tích trọng yếu có còn giá trị NaN nào không:

```python
critical_cols = [
    'Year', 'Country', 'Stage', 'Stadium', 'City', 'Home Team Name', 'Away Team Name',
    'Home Team Goals', 'Away Team Goals', 'Attendance', 'Era_Format', 'Stage_Group',
    'Goal_Margin', 'Margin_Category', 'Match_Resolution', 'Occupancy_Rate',
    'Home_Confederation', 'Away_Confederation'
]

null_summary = df[critical_cols].isna().sum().reset_index()
null_summary.columns = ['Tên cột', 'Số giá trị thiếu (NaN)']
null_summary['Tỷ lệ hoàn thiện (%)'] = 100.0 - (null_summary['Số giá trị thiếu (NaN)'] / len(df) * 100)
display(null_summary)

if null_summary['Số giá trị thiếu (NaN)'].sum() == 0:
    print('✓ XÁC NHẬN: 100% cột phân tích cốt lõi hoàn chỉnh, không có giá trị trống!')
```

## 3. Kiểm Tra Tính Duy Nhất (Uniqueness)
Đảm bảo không có bản ghi nào bị trùng lặp:

```python
dup_match_id = df.duplicated(subset=['MatchID']).sum()
dup_fixture = df.duplicated(subset=['Year', 'Datetime', 'Home Team Name', 'Away Team Name']).sum()

print(f'Số lượng trùng lặp MatchID: {dup_match_id}')
print(f'Số lượng trùng lặp lịch đấu: {dup_fixture}')
assert dup_match_id == 0, 'Cảnh báo có trùng lặp MatchID'
print('✓ XÁC NHẬN: Toàn bộ 1.068 trận đấu đều là duy nhất!')
```

## 4. Kiểm Tra Phân Bố Khán Giả & Tỷ Lệ Lấp Đầy Sân (Phục vụ TV1)
Kiểm tra số liệu Attendance và Occupancy Rate qua 4 kỷ nguyên:

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Boxplot Khán giả
sns.boxplot(ax=axes[0], data=df, x='Era_Format', y='Attendance', palette='Blues')
axes[0].set_title('Phân bố số lượng khán giả qua 4 kỷ nguyên')
axes[0].set_xlabel('Kỷ nguyên')
axes[0].set_ylabel('Khán giả / trận')
axes[0].tick_params(axis='x', rotation=15)

# Tỷ lệ lấp đầy sân trung bình
occ_era = df.groupby('Era_Format')['Occupancy_Rate'].mean().reset_index()
sns.barplot(ax=axes[1], data=occ_era, x='Era_Format', y='Occupancy_Rate', palette='viridis')
axes[1].set_title('Tỷ lệ lấp đầy sân trung bình (% Occupancy Rate)')
axes[1].set_xlabel('Kỷ nguyên')
axes[1].set_ylabel('% Lấp đầy trung bình')
axes[1].set_ylim(0, 105)
axes[1].tick_params(axis='x', rotation=15)

for p in axes[1].patches:
    axes[1].annotate(f'{p.get_height():.1f}%', (p.get_x() + p.get_width() / 2., p.get_height() - 7),
                     ha='center', va='center', color='white', fontweight='bold')

plt.tight_layout()
plt.show()
```

## 5. Kiểm Tra Chuyên Môn & Hiệu Số Bàn Thắng (Phục vụ TV2)
Kiểm tra tỷ trọng các trận Sát nút vs Hủy diệt ở Vòng bảng:

```python
group_df = df[df['Stage_Group'] == '1. Group Stage']
margin_cross = pd.crosstab(group_df['Era_Format'], group_df['Margin_Category'], normalize='index') * 100

ax = margin_cross.plot(kind='bar', stacked=True, figsize=(10, 6), colormap='Spectral', edgecolor='black')
plt.title('Tỷ trọng 3 mức khoảng cách tỷ số ở Vòng bảng qua 4 Kỷ nguyên')
plt.ylabel('Tỷ lệ phần trăm (%)')
plt.xlabel('Kỷ nguyên thể thức')
plt.legend(title='Phân loại cách biệt', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.xticks(rotation=15)
plt.tight_layout()
plt.show()

display(margin_cross.round(2))
```

## 6. Kiểm Tra Cấu Trúc Vòng Knock-out (Phục vụ TV3)
Kiểm tra tỷ lệ các kịch bản trận đấu sinh tử:

```python
ko_df = df[df['Stage_Group'].isin(['2. Early Knockout', '3. Deep Knockout'])]
ko_cross = pd.crosstab(ko_df['Era_Format'], ko_df['Match_Resolution'], normalize='index') * 100

ax = ko_cross.plot(kind='bar', stacked=True, figsize=(11, 6), colormap='plasma', edgecolor='black')
plt.title('Cấu trúc kết quả các trận Knock-out qua 4 Kỷ nguyên')
plt.ylabel('Tỷ lệ phần trăm (%)')
plt.xlabel('Kỷ nguyên thể thức')
plt.legend(title='Kết quả trận đấu', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.xticks(rotation=15)
plt.tight_layout()
plt.show()

display(ko_cross.round(2))
```

## 7. Tổng Kết & Xuất Báo Cáo
Bộ dữ liệu đã sẵn sàng 100% để chia sẻ cho nhóm!
